## ChatBedrock

Uses Converse API

```python
from langchain_aws import ChatBedrock
```

<div class="alert alert-block alert-warning">
Need to authenticate to AWS Bedrock in order to execute code below. 
</div>

```python
client = ChatBedrock(
    model_id="anthropic.claude-3-haiku-20240307-v1:0",
    model_kwargs={
        "temperature": 0,
    },
)
```

```python
multiturn_messages = [
    {
        "role": "system",
        "content": [
            {
            "text": "You are a helpful assistant that translates English to French. Translate the user sentence.",
            }
        ],
    },
    {
        "role": "human",
        "content": [
            {
            "text": "I love programming.",
            }
        ],
    }
]
```

```python
response = client.invoke(messages=multiturn_messages)
print(response.content)
```
